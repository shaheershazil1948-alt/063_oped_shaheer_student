```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import (mean_squared_error, r2_score,accuracy_score,precision_score,recall_score,
classification_report
)
df = pd.read_csv(r"C:\Users\Sherry\Desktop\opened\student.csv")
print("First 5 rows:")
print(df.head())
print("\nMissing Values:")
print(df.isnull().sum())
numeric_columns = df.select_dtypes(include=np.number).columns
for column in numeric_columns:
    df[column] = df[column].fillna(df[column].mean())
print("\nMissing values after preprocessing:")
print(df.isnull().sum())
#create
df["Pass_Fail"] = (df["Final_Marks"] >= 50).astype(int)
print("\nPass/Fail:")
print(df[["Final_Marks", "Pass_Fail"]].head())
features = ["Study_Hours", "Attendance", "Assignment_Score"]
X = df[features]
y_reg = df["Final_Marks"]
# Classification target
y_class = df["Pass_Fail"]
# Train/Test Split
X_train, X_test, y_reg_train, y_reg_test = train_test_split(
    X, y_reg, test_size=0.2, random_state=42
)
X_train_class, X_test_class, y_class_train, y_class_test = train_test_split(
    X, y_class, test_size=0.2, random_state=42
)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
X_train_class_scaled = scaler.fit_transform(X_train_class)
X_test_class_scaled = scaler.transform(X_test_class)
linear_model = LinearRegression()
linear_model.fit(X_train_scaled, y_reg_train)
y_reg_pred = linear_model.predict(X_test_scaled)
#  mean
mse = mean_squared_error(y_reg_test, y_reg_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_reg_test, y_reg_pred)

print("\n===== Linear Regression =====")
print("MSE:", mse)
print("RMSE:", rmse)
print("R² Score:", r2)

logistic_model = LogisticRegression()
logistic_model.fit(
    X_train_class_scaled,
    y_class_train
)
y_class_pred = logistic_model.predict(X_test_class_scaled)
accuracy = accuracy_score(y_class_test, y_class_pred)
precision = precision_score(y_class_test, y_class_pred)
recall = recall_score(y_class_test, y_class_pred)
print("\n===== Logistic Regression =====")
print("Accuracy:", accuracy)
print("Precision:", precision)
print("Recall:", recall)
print("classification Report:")
print(classification_report(y_class_test, y_class_pred))
```
